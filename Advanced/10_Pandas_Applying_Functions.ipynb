```python
# make sure to choose the python_course environment

# import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# we need a defined function

df_salary = df[pd.notna(df['salary_year_avg'])].copy()

def projected_salary(salary):
    return salary * 1.03

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)

df_salary[['salary_year_avg', 'salary_year_inflated']]
```

```python
# lets do an anonymous function with lambda

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)

df_salary[['salary_year_avg', 'salary_year_inflated']]
```

```python
# we need to use ast module: abstract syntax trees
# ast.literal_eval(column)

import ast

ast.literal_eval(df['job_skills'][1])
# unfortunatly, we can't use this for the whole column
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list): # notna is calling out not NaN values, i.e. values we want
        return ast.literal_eval(skill_list)

df['job_skills'] = df['job_skills'].apply(clean_list)
# we'll get an error if there are NaN values because it's expecting a string to be there
# go back up to function and pass an if statement
```

```python
# lets make the function into a lambda function

df['job_skills'] = df['job_skills'].apply(lambda skill_list: ast.literal_eval(skill_list) if pd.notna(skill_list) else skill_list)
```

```python
# lets calculate projected salary for specific rows, i.e. senior roles
# lets say senior roles' inflation is 5% while others are 3%
# paste over the function you made earlier, we'll revise it
# we'll use the axis part of apply()
# using df_salary cause it has all the salary data that's filtered

# what we had before:

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)

df_salary[['salary_year_avg', 'salary_year_inflated']]
```

```python
# so here, we're going to look at axis=1 in the df, which is the job_title_short, and we want to differentiate
# senior roles with 5% inflation and other roles as 3% inflation

def projected_salary(row):
    if "Senior" in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else:
        return 1.03 * row['salary_year_avg'] # salary year inflated would return NaN values w/o else statement

# instead of a single column, we're going to have it equal the whole dataframe 
df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

# lets check if the function worked

df_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']]
```
