```python
# make sure to choose the python_course environment

# import libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x) # ast is a module so you need to import it
```

```python
df[['job_title_short', 'job_skills']].head(5) # remember double brackets for lists
```

```python
# Transform each element of a list-like to a row, replicating index values.
    
df_exploded = df.explode('job_skills')
```

```python
# we're going to modify this
df_exploded = df.explode('job_skills')
```

```python
skills_count = df_exploded.groupby(['job_title_short', 'job_skills']).size()
```

```python
type(skills_count) # becomes a series but we'll convert to a dataframe to plot it
```

```python
df_skills_count = skills_count.reset_index(name='skill_count') # we renamed the index "0" to skill_count

df_skills_count = df_skills_count.sort_values(by='skill_count', ascending=False)

df_skills_count
```

```python
# let's change the table into something we can plot

job_title = "Data Engineer" # since we set everything up, we can insert different titles in and get charts for each
top_skills = 15

df_skills_final = df_skills_count[df_skills_count['job_title_short'] == job_title].head(top_skills)

df_skills_final.plot(kind="barh", x='job_skills', y='skill_count')
plt.gca().invert_yaxis() # invert_yaxls() is a function, make sure you include () or it won't work
plt.title(f"Top {top_skills} Skills for {job_title}s")
plt.xlabel("Job Posting Count")
plt.ylabel('')
plt.legend().set_visible(False)
plt.show()
```
