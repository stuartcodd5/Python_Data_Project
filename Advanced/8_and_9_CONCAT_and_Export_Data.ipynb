```python
# make sure to choose the python_course environment

# import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# at this point we only have one dataframe -> df

# lets make a fake dataframe

df['job_posted_month'] = df['job_posted_date'].dt.strftime('%b')

df
```

```python
# lets make a dictionary for the month dataframe we're making
# first we need a list of months

months = df['job_posted_month'].unique()

months
# we should get an array made up of months available

```

```python
# now we need to start building dictionary comprehension
# we need to display the values as part of the dataframe (i.e. Jan: df_jan)

dict_months = {month: df[df['job_posted_month'] == month] for month in months}
```

```python
dict_months['Apr'] # use abbreviated month names
```

```python
# lets get the 2023 Q1 data
# now we use concat()

df_q1 = pd.concat([dict_months['Jan'], dict_months['Feb'], dict_months['Mar']], ignore_index = True)
# we aren't completely sure it worked, so lets make it a variable and check
```

```python
df_q1['job_posted_month'].value_counts().plot(kind="bar")
```
