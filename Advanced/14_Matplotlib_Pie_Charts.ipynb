```python
# make sure to choose the python_course environment

# import libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
# we're going to start with job_work_from_home (Boolean expression)

df['job_work_from_home'].value_counts().plot(kind="pie", startangle=90, autopct='%1.1f%%') # autopct % is showing the pcts. for the variables
plt.title("WFH Status")
plt.ylabel("")
plt.show()

# the f in autopct specifies that it's a floating point number
```

```python
fig, ax = plt.subplots(1, 3)

dict_column = {
    'job_work_from_home': "Work From Home",
    'job_no_degree_mention': "Job Degree Req.",
    'job_health_insurance': "Health Insurance Offered"
}

for i, (column, title) in enumerate(dict_column.items()): # column, titie is pulling from the tuple, must be in ()
    ax[i].pie(df[column].value_counts(), startangle=90, autopct='%1.1f%%', labels=['False', 'True'])
    ax[i].set_title(title)

plt.show()
```
