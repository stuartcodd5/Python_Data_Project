```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# load the data
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset['train'].to_pandas()

# clean
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# we're going to look for median salaries by job title
# aggregate the job titles

job_salary = df.groupby('job_title_short')['salary_year_avg'].agg("median").sort_values() # median would be the best agg to analyze salaries

```

```python
job_salary.plot(kind="barh")
plt.ylabel("")
plt.xlabel("Salary (In USD)")
plt.title("Median Salary by Job Title")
plt.show()
```
